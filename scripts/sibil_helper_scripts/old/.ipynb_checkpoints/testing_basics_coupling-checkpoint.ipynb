```python
import basics
import matplotlib.pyplot as plt
import numpy as np
from boututils.datafile import DataFile
from pathlib import Path
```

```python
def get_bgrid_path(bout_run_dir):
    with open(bout_run_dir / "BOUT.inp") as f:
        for line in f:
            if line.strip().startswith("grid"):
                start, grid = line.split("=")
                grid = grid.strip()
                grid = grid.strip('\"')
                return grid
```

```python
def get_bgrid(bout_run_dir):
    bgrid_name = get_bgrid_path(bout_run_dir)
    bgrid_path = bout_run_dir / bgrid_name
    return DataFile(str(bgrid_path))
```

```python
bgrid_loc = Path("/global/cfs/projectdirs/m4491/btaczak/SIBIL-runs/DIIID_LSN_1D_midplane_without_drifts/Shelf_new_grid2_sinks2_gradfix")
grid = get_bgrid(bgrid_loc)
for key in grid.keys():
    shape = grid[key].shape
    if len(shape) == 2:# and key == "Jpar0":
        print(f"{key}: {shape}")
        basics.plot_matrix(grid[key], key,dpi=100)
    # elif len(shape) == 1:
    #     for i,val in enumerate(grid[key]):
    #         print(f"{key}, {i}: {val}")
    # elif len(shape) == 0:
    #     print(f"{key}: {grid[key]}")
        
```
